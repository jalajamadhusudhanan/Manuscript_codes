```python
import matplotlib.animation as animation
from matplotlib.animation import FFMpegWriter

# Load the kymograph data
kymo_path=r"Z:\jalaja\behaviour\Autoscope\data\27_11_22_w2\2022-11-27_13-19_w2_Ch0\skeleton_spline_K.csv"
df_kymo = pd.read_csv(kymo_path, header=None)

# Load the video file
video_path = r"Z:\jalaja\behaviour\Autoscope\videos\overlaid_2022-11-27_13-19_w2_Ch0_MMStack_hit_1620.avi"  # Path to the input video file
cap = cv2.VideoCapture(video_path)

# Get the total number of frames
num_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))

# Get video properties
fps = cap.get(cv2.CAP_PROP_FPS)
frame_width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
frame_height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))

# Create the figure and two subplots (one for video, one for kymograph)
fig, (ax_video, ax_kymo) = plt.subplots(2, 1, figsize=(10, 8))

# Define the kymograph plot extent
extent = [0, df_kymo.shape[0] / 5, df_kymo.shape[1], 0]

# Display the kymograph in the lower subplot
im = ax_kymo.imshow(df_kymo.T.loc[:, 16200:16440], origin="upper", cmap='Spectral', extent=extent, aspect=10, vmin=-0.01, vmax=0.01)

# Add colorbar for kymograph
cbar = fig.colorbar(im, ax=ax_kymo, orientation='vertical')
cbar.set_label('curvature(°)')

# Initialize the vertical line (slider)
line = ax_kymo.axvline(x=0, color='k', linestyle='--')

# Set labels and limits for the kymograph subplot
ax_kymo.set_xlabel('Time (s)')
ax_kymo.set_ylabel('#body segment')
# Set custom tick labels for x-axis
ticks = np.linspace(0, df_kymo.shape[0] / 5, num=5)  # Generate tick positions
tick_labels = np.linspace(-7, 17, num=5)  # Generate tick labels
ax_kymo.set_xticks(ticks)
ax_kymo.set_xticklabels(np.round(tick_labels, 2))

# Hide video subplot axes initially (they will be updated in the animation function)
ax_video.axis('off')

# Function to update the plot and video frame
def animate(i):
    ret, frame = cap.read()
    if not ret:
        return None

    # Clear previous frames in the video subplot
    ax_video.cla()

    # Display the video frame in the upper subplot
    frame = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
    ax_video.imshow(frame)
    ax_video.axis('off')  # Hide axes

    # Update the vertical line in the kymograph subplot
    line.set_xdata(i * (df_kymo.shape[0] / 5) / num_frames)

    return line,

# Initialize the writer
writer = FFMpegWriter(fps=fps)

# Create the animation
ani = animation.FuncAnimation(fig, animate, frames=num_frames, blit=False)

# Save the animation
ani.save(r"Z:\jalaja\behaviour\Autoscope\videos\kymograph_hit_animation.mp4", writer=writer)

# Release the video capture
cap.release()
```
