```python
import math
import pandas as pd
import warnings
warnings.filterwarnings("ignore")
import numpy as np
import os
from glob import glob
import matplotlib.pyplot as plt
```

```python
# Apply smoothing to each row
def smooth_row(row, window_size=3):
    smoothed_row = row.rolling(window=window_size, min_periods=1).mean()
    return smoothed_row

def unit_vector(vector):
    """ Returns the unit vector of the vector.  """
    return vector / np.linalg.norm(vector)

def angle_between(v1, v2):
    """ Returns the angle in radians between vectors 'v1' and 'v2'::

            >>> angle_between((1, 0, 0), (0, 1, 0))
            1.5707963267948966
            >>> angle_between((1, 0, 0), (1, 0, 0))
            0.0
            >>> angle_between((1, 0, 0), (-1, 0, 0))
            3.141592653589793
    """
    v1_u = unit_vector(v1)
    v2_u = unit_vector(v2)
    return np.arccos(np.clip(np.dot(v1_u, v2_u), -1.0, 1.0))
def angle(p0, p1, p2):
    v0 = np.array(p0) - np.array(p1)
    v1 = np.array(p2) - np.array(p1)
    angle = np.math.atan2(np.linalg.det([v0,v1]),np.dot(v0,v1))
    #print(v0,v1,angle)
    return np.degrees(angle)
    
#both atan2 and arccos gives same angle values checked!tan2 is better
    # cos_angle = np.dot(v0, v1) / (np.linalg.norm(v0) * np.linalg.norm(v1))
    # if cos_angle < -1.0:
    #     cos_angle = -1.0
    # elif cos_angle > 1.0:
    #     cos_angle = 1.0
    # angle_rad = np.arccos(cos_angle)
    # angle_deg = np.degrees(angle_rad)
    #print(cos_angle,angle_rad,angle_deg)
    # return angle_deg
def calculate_speed(xy):
    #https://codereview.stackexchange.com/questions/158688/calculating-speed-from-a-pandas-dataframe-with-time-x-and-y-columns
    movements_over_timesteps = (
        np.roll(xy, -1, axis=0)
        - xy)[:-1]

    speeds = np.sqrt(
        movements_over_timesteps.X ** 2 +
        movements_over_timesteps.Y ** 2
    ) / movements_over_timesteps['time']

    return pd.DataFrame({
        'time': xy['time'][:-1],
        'Speed': speeds,
    })
    
```

Roam-dwell analysis, based on the definition of Ben Arous et al, 2009 

```python
#binning
bin_size = 10  # in seconds
bin_frame = bin_size * 10

# Create a new dataframe to store binned values
binned_df_speed = pd.DataFrame(index=df_speed.index, columns=range(df_speed.shape[1] // bin_frame - 1))

# Iterate over each row and calculate binned mean
for i in range(df_speed.shape[0]):
    for j in range(df_speed.shape[1] // bin_frame - 1):
        start_idx = j * bin_frame
        end_idx = (j + 1) * bin_frame
        binned_df_speed.iloc[i, j] = df_speed.iloc[i, start_idx:end_idx].mean()

print(binned_df_speed)
        
```

```python
#binning
bin_size = 10  # in seconds
bin_frame = bin_size * 10

# Create a new dataframe to store binned values
binned_angular_speed = pd.DataFrame(index=df_angle.index, columns=range(df_angle.shape[1] // bin_frame - 1))

# Iterate over each row and calculate binned mean
for i in range(df_angle.shape[0]):
    for j in range(df_angle.shape[1] // bin_frame - 1):
        start_idx = j * bin_frame
        end_idx = (j + 1) * bin_frame
        binned_angular_speed.iloc[i, j] = angular.iloc[i, start_idx:end_idx].abs().mean()

print(binned_angular_speed)
```

```python
binary_vector = (binned_speed_data_filtered >= angular_data_filtered / 150).astype(int)
print(binary_vector)
```

```python
from scipy.interpolate import interp1d

# Initialize an empty pandas DataFrame to hold the interpolated values
binary_interpolated_vector = pd.DataFrame(index=range(binary_vector.shape[0]), columns=range(18900))

# Iterate over the rows of the binary_vector
for i in range(binary_vector.shape[0]):
    original_length = binary_vector.shape[1]
    
    # Create an array of original indices (0 to original_length - 1)
    original_indices = np.arange(original_length)
    
    # Create an array of new indices (0 to 18900 - 1) for interpolation
    new_indices = np.linspace(0, original_length - 1, 18900)
    
    # Perform linear interpolation using scipy's interp1d
    interpolator = interp1d(original_indices, binary_vector.iloc[i,:], kind='linear')
    
    # Interpolate to get new values
    interpolated_values = interpolator(new_indices)
    
    # Apply a threshold to ensure the values remain binary (0s and 1s)
    binary_interpolated_vector.loc[i, :] = (interpolated_values >= 0.5).astype(int)

# The binary_interpolated_vector DataFrame now holds the interpolated values
print(binary_interpolated_vector)
```

```python
# plot roam-dwell binary vector for individual worms, 
#conclusion: mostly worms roam, rarely dwell

def plot_multiple_subplots(df1, df2, df3, save_folder=r"C:\Users\Jalaja Madhusudhanan\Desktop\autoscope_stage position\roam_dwell_plots", nrows=25):
    """
    Plot 3 subplots for each row in the given dataframes.
    
    Parameters:
    df1, df2, df3 : pd.DataFrame
        Dataframes to plot (with equal number of rows).
    save_folder : str
        The folder to save plots.
    nrows : int
        The number of rows to loop through (default is 117).
    """
    for i in range(nrows):
        fig, axs = plt.subplots(3, 1, figsize=(10, 8), sharex=True)  # 3 subplots, 1 column
        
        # Plot row from df1 on the first subplot
        axs[0].plot(df1.iloc[i, :-1], label='DF1', color='blue')
        axs[0].set_title(f'Dataframe 1 - Row {i}')
        axs[0].set_ylabel('Value')
        axs[0].legend()

        # Plot row from df2 on the second subplot
        axs[1].plot(df2.iloc[i, :-1], label='DF2', color='green')
        axs[1].set_title(f'Dataframe 2 - Row {i}')
        axs[1].set_ylabel('Value')
        axs[1].legend()

        # Plot row from df3 on the third subplot
        axs[2].plot(df3.iloc[i, :-1], label='DF3', color='red')
        axs[2].set_title(f'Dataframe 3 - Row {i}')
        axs[2].set_xlabel('Frame')
        axs[2].set_ylabel('Value')
        axs[2].legend()

        plt.tight_layout()
        
        # Save plot for each row
        plt.savefig(f"{save_folder}/plot_row_{i}.png")
        plt.close()
```

```python
# Example usage:
plot_multiple_subplots(Angular_speed, df_speed, binary_interpolated_vector)
```
